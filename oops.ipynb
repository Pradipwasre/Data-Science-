```python
class Car: # partent Class
    def __init__(self , brand , model):
        self.brand = brand
        self.model = model
    
    # method
    def info(self):
        return f"Brand Name : {self.brand} | Model Name: {self.model}"
    

class ElectricCar(Car): # Child Class 1 
    def __init__(self, brand , model , Battary_size):
        super().__init__(brand , model)
        self.Battary_size = Battary_size

    def country(self):
        return "USA"
    
class HybridCar(Car): # Child class 2 
    def __init__(self, brand, model, fule_type, country):
        super().__init__(brand , model)
        self.fuel_type = fule_type
        self.country = country
    
    #def country(self):
    #    return self.country
```
