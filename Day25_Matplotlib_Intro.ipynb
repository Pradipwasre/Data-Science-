## 1. Getting Started with Matplotlib

**What is Matplotlib?**
Matplotlib is the most widely used Python library for Data Visualization. It lets us turn raw numbers (like our Blinkit stock and sales data) into charts that are easy to understand at a glance.

**Why do we use it for Data Visualization?**
- A manager cannot read 9000+ rows of a CSV, but he can read one bar chart in 5 seconds
- It helps spot patterns, trends, and outliers (e.g. which warehouse is low on stock)
- It is the foundation library — Seaborn, Pandas `.plot()`, and many other tools are built on top of it

## 2. Types of Plots

We will cover 5 core plot types, each with a Blinkit business question attached.

### 2.1 Line Plot
**Use case:** plotting continuous data over time or sequence.

**Business question:** How did revenue move across the first 30 sale records of WH001 (Pune Hinjewadi)?

````python
wh1 = df[df["warehouse_id"] == "WH001"].head(30)

plt.figure(figsize=(10, 5))
plt.plot(range(len(wh1)), wh1["revenue_inr"])
plt.title("Revenue Trend - First 30 Records (WH001)")
plt.xlabel("Record Number")
plt.ylabel("Revenue (INR)")
plt.tight_layout()
plt.show()
````

### 2.2 Bar Plot
**Use case:** comparing categories.

**Business question:** Which product category sells the most units overall?

(No `groupby` yet — we build the totals with a simple loop and boolean filtering.)

````python
categories_list = df["category"].unique()
total_units = []

for cat in categories_list:
    cat_data = df[df["category"] == cat]          # boolean filtering
    total_units.append(cat_data["units_sold"].sum())

plt.figure(figsize=(10, 5))
plt.bar(categories_list, total_units)
plt.title("Total Units Sold by Category")
plt.xlabel("Category")
plt.ylabel("Units Sold")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
````

### 2.3 Histogram
**Use case:** visualizing distribution of data.

**Business question:** How is the delivery time (in minutes) distributed across all orders?

````python
plt.figure(figsize=(8, 5))
plt.hist(df["delivery_time_mins"], bins=20, edgecolor="black")
plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Frequency")
plt.show()
````

### 2.4 Box Plot
**Use case:** detecting spread and outliers.

**Business question:** How does unit price vary across categories, and are there any outlier priced products?

````python
categories = df["category"].unique()
data_by_category = [df[df["category"] == cat]["unit_price_inr"] for cat in categories]

plt.figure(figsize=(10, 5))
plt.boxplot(data_by_category, tick_labels=categories)
plt.title("Unit Price Spread by Category")
plt.xlabel("Category")
plt.ylabel("Unit Price (INR)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
````

### 2.5 Scatter Plot
**Use case:** relationship between two variables.

**Business question:** Is there a relationship between units sold and revenue generated?

````python
plt.figure(figsize=(8, 5))
plt.scatter(df["units_sold"], df["revenue_inr"], alpha=0.4)
plt.title("Units Sold vs Revenue")
plt.xlabel("Units Sold")
plt.ylabel("Revenue (INR)")
plt.show()
````

## 3. Plot Customization

A plot without labels is not useful in a real report. Let's customize the same category-sales bar chart step by step.

**Adding titles, axis labels, and legends**

````python
plt.figure(figsize=(10, 5))
plt.bar(categories_list, total_units, label="Units Sold")
plt.title("Category-wise Units Sold - Blinkit Warehouse", fontsize=14)
plt.xlabel("Category", fontsize=11)
plt.ylabel("Total Units Sold", fontsize=11)
plt.legend()
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
````

**Changing colors, markers, and line styles**

Common style options:
- `color` : `"red"`, `"green"`, `"#1f77b4"`, etc.
- `marker` : `"o"` circle, `"s"` square, `"^"` triangle, `"*"` star
- `linestyle` : `"-"` solid, `"--"` dashed, `":"` dotted, `"-."` dash-dot

````python
storage_types = df["storage_type"].unique()
avg_delivery = []

for stype in storage_types:
    stype_data = df[df["storage_type"] == stype]   # boolean filtering, no groupby
    avg_delivery.append(stype_data["delivery_time_mins"].mean())

plt.figure(figsize=(8, 5))
plt.plot(storage_types, avg_delivery,
         color="green", marker="o", linestyle="--", linewidth=2, markersize=8)
plt.title("Average Delivery Time by Storage Type")
plt.xlabel("Storage Type")
plt.ylabel("Avg Delivery Time (mins)")
plt.grid(True, alpha=0.3)
plt.show()
````

**Comparing two lines with a legend — In Stock vs Low Stock count per city**

````python
cities_list = df["city"].unique()
in_stock_counts = []
low_stock_counts = []

for city in cities_list:
    city_data = df[df["city"] == city]              # boolean filtering, no groupby
    in_stock_counts.append((city_data["availability_status"] == "In Stock").sum())
    low_stock_counts.append((city_data["availability_status"] == "Low Stock").sum())

plt.figure(figsize=(10, 5))
plt.plot(cities_list, in_stock_counts, color="blue", marker="o", label="In Stock")
plt.plot(cities_list, low_stock_counts, color="red", marker="s", label="Low Stock")
plt.title("In Stock vs Low Stock Records by City")
plt.xlabel("City")
plt.ylabel("Number of Records")
plt.legend()
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
````

**Saving plots as images (.png, .jpg)**

Use `plt.savefig()` **before** `plt.show()` (calling it after `show()` in some environments saves a blank image).
`dpi` controls image quality, higher dpi means a sharper image for reports and presentations.

````python
plt.figure(figsize=(10, 5))
plt.bar(categories_list, total_units, color="orange")
plt.title("Category-wise Units Sold - Blinkit Warehouse")
plt.xlabel("Category")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig("category_sales.png", dpi=300)
plt.savefig("category_sales.jpg", dpi=300)

plt.show()
print("Saved as category_sales.png and category_sales.jpg")
````
