## 1. Loading CSV Files

Let's start by loading our dataset using `pd.read_csv()`. This is the most common way to bring external data into a Pandas DataFrame.


````python
import pandas as pd
import numpy as np

# Basic syntax: pd.read_csv(filepath)
df = pd.read_csv("myntra_student_shopping_data.csv")
print("Data loaded successfully!")
print(type(df))
````

### Specifying file path and delimiter

By default, `pd.read_csv()` assumes the file is comma-separated (`,`). If your file uses a different delimiter (like `;` or tab `\t`), you specify it using the `sep` or `delimiter` parameter.

```python
# Example if it were semicolon-separated:
df = pd.read_csv("myntra_student_shopping_data.csv", sep=';')

# Example for tab-separated file:
df = pd.read_csv("data.tsv", delimiter='\t')

# You can also load from a specific folder path:
df = pd.read_csv("/mnt/user-data/uploads/myntra_student_shopping_data.csv")
```

Since our file is a standard CSV (comma-separated), the default works fine.

## 3. Basic Data Manipulation (EDA)

Now let's manipulate the data — selecting, filtering, sorting, renaming, and handling missing values.


### Selecting columns (`df['col']`)

````python
# Selecting a single column returns a Series
df['Product_Name'].head()
````

````python
# Selecting multiple columns returns a DataFrame (note the double brackets)
df[['Product_Name', 'Brand', 'Final_Price']].head()
````

### Selecting rows with `.loc[]` and `.iloc[]`

- `.loc[]` → label/condition-based selection
- `.iloc[]` → position/index-based selection


````python
# .loc[] with row label (here index is default integer, so it behaves like position too)
df.loc[0]
````

````python
# .loc[] to select specific rows AND columns
df.loc[0:4, ['Product_Name', 'Category', 'Final_Price']]
````

````python
# .iloc[] selects by integer position (row 0 to 4, columns 0 to 3)
df.iloc[0:5, 0:4]
````

````python
# .iloc[] single cell access: row 2, column 3
df.iloc[2, 3]
````

### Filtering rows with conditions

````python
# Filter: products priced above 2000
expensive_items = df[df['Price'] > 2000]
print("Number of items priced above 2000:", expensive_items.shape[0])
expensive_items.head()
````

````python
# Multiple conditions: Footwear category AND rating above 4
good_shoes = df[(df['Category'] == 'Footwear') & (df['Rating'] > 4)]
good_shoes[['Product_Name', 'Brand', 'Rating', 'Final_Price']].head()
````

````python
# Filter using .isin() — useful for multiple category matches
student_essentials = df[df['Category'].isin(['Bags', 'Footwear'])]
print("Bags & Footwear orders:", student_essentials.shape[0])
student_essentials.head()
````

### Sorting data with `.sort_values()`

````python
# Sort by Final_Price ascending (default)
df.sort_values('Final_Price').head()
````

````python
# Sort descending — most expensive items first
df.sort_values('Final_Price', ascending=False).head()
````

````python
# Sort by multiple columns: Category (A-Z), then Rating (high to low)
df.sort_values(['Category', 'Rating'], ascending=[True, False]).head(10)
````

### Renaming columns with `.rename()`

````python
# Rename specific columns using a dictionary — doesn't affect original unless we reassign or use inplace=True
df_renamed = df.rename(columns={'Final_Price': 'Price_After_Discount', 'Number_of_Ratings': 'Total_Ratings'})
df_renamed.columns
````

### Checking for missing values (`.isnull()`, `.notnull()`)

````python
# .isnull() returns True/False for each cell
df.isnull().head()
````

````python
# Much more useful: count missing values per column
df.isnull().sum()
````

````python
# .notnull() is the opposite — True where data EXISTS
df.notnull().sum()
````

````python
# Percentage of missing values per column
(df.isnull().sum() / len(df) * 100).round(2)
````

### Dropping missing values (`.dropna()`) and filling them (`.fillna()`)

````python
# dropna() removes ANY row that has at least one missing value
df_dropped = df.dropna()
print("Original shape:", df.shape)
print("Shape after dropna():", df_dropped.shape)
````

````python
# dropna with subset — only drop rows missing in specific important columns
df_dropped_subset = df.dropna(subset=['City'])
print("Shape after dropping rows with missing City:", df_dropped_subset.shape)
````

````python
# fillna() — fill missing Rating with the average rating instead of dropping data
avg_rating = df['Rating'].mean()
df_filled = df.copy()
df_filled['Rating'] = df_filled['Rating'].fillna(round(avg_rating, 1))

# fill missing City with 'Unknown'
df_filled['City'] = df_filled['City'].fillna('Unknown')

# fill missing Delivery_Days with the median delivery time
df_filled['Delivery_Days'] = df_filled['Delivery_Days'].fillna(df_filled['Delivery_Days'].median())

print("Missing values after fillna():")
df_filled.isnull().sum()
````

> **Why `fillna()` over `dropna()` here?** Dropping rows loses valuable order data (price, brand, category etc.) just because one field like `Rating` is missing. Filling with a sensible default (mean/median/placeholder) preserves the dataset size for further analysis — a common real-world trade-off.


## 4. Saving DataFrame to CSV with `.to_csv()`

After cleaning, we usually save our processed data for future use.


````python
# Save the cleaned dataset (missing values handled) to a new CSV file
df_filled.to_csv("myntra_student_shopping_data_cleaned.csv", index=False)
print("Cleaned file saved as 'myntra_student_shopping_data_cleaned.csv'")
````

````python
# Quick verification — read it back
check_df = pd.read_csv("myntra_student_shopping_data_cleaned.csv")
check_df.isnull().sum()
````
