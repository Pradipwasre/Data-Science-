# Day 26 - Introduction to Seaborn
Dataset used: **Google Trends 2026** (search interest across 6 categories, month by month)

## What is Seaborn and why use it over Matplotlib
- Seaborn is built on top of Matplotlib, made for statistical plots
- Needs less code than Matplotlib for the same chart
- Works directly with pandas DataFrames
- Has better default colors, styles and themes

```
# Installing Seaborn (run once in terminal / notebook)
# pip install seaborn
```

```
# Importing Seaborn
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Loading the dataset
- Using our own CSV file (`google_trends_2026.csv`) instead of a built-in dataset
- `sns.load_dataset()` is used only for Seaborn's sample datasets like `tips`, `iris`
- For real projects, we load our own CSV using pandas

```
df = pd.read_csv("google_trends_2026.csv")
df.head()
```

### Bar Plot - sns.barplot()
- Used to compare a number across different categories
- Here we compare average search interest of each category in 2026

```
sns.barplot(data=df, x="Category", y="Search_Interest")
plt.show()
```

### Line Plot - sns.lineplot()
- Used to show a trend over time
- Here we see how AI Tools search interest changed month by month in 2026

```
ai_df = df[df["Category"] == "AI Tools"]
sns.lineplot(data=ai_df, x="Month", y="Search_Interest")
plt.xticks(rotation=45)
plt.show()
```

### Histogram - sns.histplot()
- Used to see how values are distributed (spread out)
- Here we see the distribution of Search_Interest values across all categories

```
sns.histplot(data=df, x="Search_Interest", bins=10)
plt.show()
```

### Scatter Plot - sns.scatterplot()
- Used to see the relationship between two numeric columns
- Here we check if higher Search_Interest also means higher Engagement_Score

```
sns.scatterplot(data=df, x="Search_Interest", y="Engagement_Score", hue="Category")
plt.show()
```

### Box Plot - sns.boxplot()
- Used to see spread, median and outliers of a value across categories
- Here we compare Search_Interest spread for each category

```
sns.boxplot(data=df, x="Category", y="Search_Interest")
plt.xticks(rotation=45)
plt.show()
```

### Pair Plot - sns.pairplot()
- Used to explore relationships between multiple numeric variables at once
- Plots every column against every other column in one go
- Here we compare Search_Interest, Engagement_Score and Ad_Spend_Index together

```
sns.pairplot(df[["Search_Interest", "Engagement_Score", "Ad_Spend_Index", "Category"]], hue="Category")
plt.show()
```

## Customization
### Changing themes and styles
- `sns.set_theme()` changes the overall look (background, grid, font size)
- `sns.set_style()` changes just the background style (darkgrid, whitegrid, ticks, etc.)

```
sns.set_theme(style="whitegrid")
sns.barplot(data=df, x="Category", y="Search_Interest")
plt.show()
```

### Adding titles and axis labels
- Makes the chart easier to read and understand for anyone viewing it

```
sns.lineplot(data=ai_df, x="Month", y="Search_Interest")
plt.title("AI Tools - Google Search Interest in 2026")
plt.xlabel("Month")
plt.ylabel("Search Interest (0-100)")
plt.xticks(rotation=45)
plt.show()
```

### Adjusting colors and palettes
- `palette` parameter changes the color scheme of the plot
- Useful palettes: `"pastel"`, `"deep"`, `"muted"`, `"dark"`, `"Set2"`

```
sns.barplot(data=df, x="Category", y="Search_Interest", palette="Set2")
plt.title("Average Search Interest by Category (2026)")
plt.show()
```
