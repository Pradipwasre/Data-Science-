## Different Ways to Create Arrays
NumPy provides multiple functions to create arrays efficiently.

### Using np.array() 

```python
import numpy as np

arr_list = np.array([1, 2, 3, 4, 5])
arr_tuple = np.array((10, 20, 30))
```

```python
print("Array from list:", arr_list)
print("Array from tuple:", arr_tuple)
```

### Using np.zeros() and np.ones()

```python
zeros_arr = np.zeros((2,3))
ones_arr = np.ones((2,3))
```

```python
print("Zeros Array:\n", zeros_arr)
print("Ones Array:\n", ones_arr)
## Using np.arange()
```

## Using np.arange()

```python
arr_range = np.arange(0, 10, 2)
```

```python
print("Array using arange:", arr_range)
```

### Using np.linspace()

```python
arr_linspace = np.linspace(0, 1, 5)
```

```python
print("Array using linspace:", arr_linspace)
```

### Using np.random

```python
rand_arr = np.random.rand(2,3)
randn_arr = np.random.randn(2,3)
randint_arr = np.random.randint(1, 10, (2,3))
```

```python
print("Random Uniform:\n", rand_arr)
```

```python
print("Random Normal:\n", randn_arr)
```

```python
print("Random Integers:\n", randint_arr)
```

### Slicing in Arrays

#### Slicing allows extracting parts of arrays using the `:` operator.

```python
arr = np.arange(10)
```

```python
print("Original Array:", arr)
```

```python
print("Slice [2:7]:", arr[2:7])
```

### Slicing in 1D Arrays

```python
arr = np.arange(10)
```

```python
print("First 5 elements:", arr[:5])
```

```python
print("Last 3 elements:", arr[-3:])
```

### Slicing in 2D Arrays

```python
arr2d = np.array([[10,20,30],[40,50,60],[70,80,90]])
print("Original 2D Array:\n", arr2d)
```

```python
print("First row:", arr2d[0])
print("First column:", arr2d[:,0])
print("Subarray:\n", arr2d[0:2, 1:3])
```

### Step Size in Slicing

```python
arr = np.arange(20)
```

```python
print("Array:", arr)
```

```python
print("Slice with step [::2]:", arr[::2])
```
