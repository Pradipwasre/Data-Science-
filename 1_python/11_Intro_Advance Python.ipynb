## Installing Anaconda
Anaconda is a distribution of Python and R for scientific computing.  
It simplifies package management and deployment, especially for **Data Science** and **Machine Learning** projects.

## Why use Anaconda for Data Science?
- Comes with 1,500+ pre-installed packages (NumPy, Pandas, Matplotlib, Scikit-learn, etc.)
- Includes **Jupyter Notebook** for interactive coding
- Easy environment management
- Cross-platform (Windows, macOS, Linux)

## Downloading and Installing Anaconda
Steps:
1. Visit https://www.anaconda.com/download
2. Choose installer for your OS (Windows, macOS, Linux)
3. Run installer and follow setup instructions
4. Verify installation using:
   conda --version

## Navigating Anaconda Navigator
- GUI tool to manage environments and packages
- Launch Jupyter Notebook, Spyder, VS Code
- Install/Update packages without terminal

## Creating and Managing Virtual Environments
Virtual environments help isolate dependencies for different projects.

#### Create new environment
conda create -n myenv python=3.9

#### Activate environment
conda activate myenv

#### Deactivate environment
conda deactivate

## Introduction to NumPy
NumPy = Numerical Python  
- Provides support for **multi-dimensional arrays**  
- Optimized for numerical computations  
- Essential for Data Science, ML, and Deep Learning

## Installing NumPy

```python
pip install numpy
```

```python
import numpy as np
print("NumPy version:", np.__version__)
```

```python
# 1D Array
arr1 = np.array([1, 2, 3, 4, 5])
print("1D Array:", arr1)
```

```python
# 2D Array
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print("2D Array:\n", arr2)
```

### Array Attributes

```python
print("Shape:", arr2.shape)
print("Data Type:", arr2.dtype)
print("Size:", arr2.size)
print("Dimensions:", arr2.ndim)
```

### Type Conversion

```python
arr_float = arr1.astype(float)
print("Converted to float:", arr_float)
print("Data Type:", arr_float.dtype)
```

### Simple Array Operations

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
```

```python
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

### Accessing & Slicing Arrays 

```python
print("First element:", arr1[0])
print("Slice [1:4]:", arr1[1:4])
```

```python
print("Row 1:", arr2[0])
print("Element [1,2]:", arr2[1,2])
```

### Grayscale Image with NumPy + Matplotlib

```python
import matplotlib.pyplot as plt

# Create a 5x5 array with values from 0 to 255

image = np.array([[0, 50, 100, 150, 200],
                  [25, 75, 125, 175, 225],
                  [50, 100, 150, 200, 250],
                  [75, 125, 175, 225, 255],
                  [100, 150, 200, 250, 255]])
```

```python
plt.imshow(image, cmap='gray')
plt.colorbar()
plt.title("Simple Grayscale Image")
plt.show()
```
