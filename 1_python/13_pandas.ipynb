## What is Pandas?
Pandas is a Python library for **data analysis and manipulation**.  
It provides:
- Fast and flexible data structures
- Easy handling of structured/tabular data
- Integration with NumPy and other libraries

## Pandas Data Structures
- **Series** → One-dimensional labeled array (like a column).  
- **DataFrame** → Two-dimensional labeled data structure (rows and columns).


## Creating a Series
A Series can be created from:
- Python list
- NumPy array
- Dictionary

```python
import numpy as np
import pandas as pd

s_list = pd.Series([10, 20, 30, 40])
print(s_list)
```

### From a NumPy Array

```python
arr = np.array([1, 2, 3, 4, 5])
s_array = pd.Series(arr)
print(s_array)
```

### From a Dictionary

```python
data = {"a": 100, "b": 200, "c": 300}
s_dict = pd.Series(data)
print(s_dict)
```

### From a List of Dictionaries

```python
data = [{"name": "Alice", "age": 25}, {"name": "Bob", "age": 30}]
df_list_dict = pd.DataFrame(data)
print(df_list_dict)
```

### From a NumPy Array

```python
arr = np.array([[1,2,3],[4,5,6]])
df_array = pd.DataFrame(arr, columns=["A","B","C"])
```

```python
print(df_array)
```
