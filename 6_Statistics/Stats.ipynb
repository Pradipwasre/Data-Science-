# Small Standard Deviation → Predictable Deliveries

- When SD is **small**, delivery times are tightly clustered around the mean.
- Customers can expect consistent delivery times (e.g., ~30 minutes).
- Business implication: Blinkit can confidently promise reliable service.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Small SD dataset (predictable deliveries)
small_sd_data = np.array([29, 30, 31, 30, 32, 31, 30, 29, 32, 30, 31, 30, 29, 31, 32])

# Calculate mean, variance, SD
mean_small = np.mean(small_sd_data)
variance_small = np.var(small_sd_data)
sd_small = np.std(small_sd_data)

print("Mean:", mean_small)
print("Variance:", variance_small)
print("Standard Deviation:", sd_small)

# Plot histogram
plt.figure(figsize=(6,4))
sns.histplot(small_sd_data, bins=6, kde=True, color="green")
plt.title(f"Small SD Deliveries\nMean={mean_small:.2f}, SD={sd_small:.2f}")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Frequency")
plt.show()
```

# Large Standard Deviation → Inconsistent Deliveries

- When SD is **large**, delivery times vary widely.
- Customers face unpredictability (some deliveries very fast, others very late).
- Business implication: Blinkit must investigate causes (traffic, prep delays, rider allocation) and improve operations.

```python
# Large SD dataset (inconsistent deliveries)
large_sd_data = np.array([20, 22, 25, 28, 30, 35, 40, 45, 50, 55, 60, 42, 48, 25, 60])

# Calculate mean, variance, SD
mean_large = np.mean(large_sd_data)
variance_large = np.var(large_sd_data)
sd_large = np.std(large_sd_data)

print("Mean:", mean_large)
print("Variance:", variance_large)
print("Standard Deviation:", sd_large)

# Plot histogram
plt.figure(figsize=(6,4))
sns.histplot(large_sd_data, bins=8, kde=True, color="red")
plt.title(f"Large SD Deliveries\nMean={mean_large:.2f}, SD={sd_large:.2f}")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Frequency")
plt.show()
```

# Skewness and Kurtosis in Statistics

## Skewness
- **Positive Skewness**: Data has a long right tail (e.g., delivery times mostly fast, but a few extreme delays).
- **Negative Skewness**: Data has a long left tail (e.g., exam scores mostly high, but a few very low scores).
- **Zero Skewness**: Data is symmetric (mean ≈ median ≈ mode).

## Kurtosis
- **Mesokurtic**: Normal distribution (moderate peak, normal tails).
- **Leptokurtic**: Sharp peak, heavy tails (more extreme outliers).
- **Platykurtic**: Flat peak, light tails (fewer extremes).

These measures help companies like Blinkit detect **bias** (skewness) and **extreme risks** (kurtosis) in delivery times.


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis

# Example datasets
positive_skew = [20, 22, 25, 30, 60]   # Delivery times with extreme delay
negative_skew = [10, 40, 45, 48, 50]   # Exam scores with few very low marks
normal_data   = [28, 30, 32, 34, 36]   # Symmetric delivery times

datasets = {
    "Positive Skewness": positive_skew,
    "Negative Skewness": negative_skew,
    "Normal (Mesokurtic)": normal_data
}

# Plot histograms
plt.figure(figsize=(15, 4))
for i, (title, data) in enumerate(datasets.items(), 1):
    plt.subplot(1, 3, i)
    sns.histplot(data, bins=5, kde=True, color="skyblue")
    plt.title(f"{title}\nSkew={skew(data):.2f}, Kurtosis={kurtosis(data):.2f}")
plt.tight_layout()
plt.show()
```

# Quartiles and Interquartile Range (IQR)

## Step 1: Arrange Data in Ascending Order
Example dataset (Blinkit delivery times in minutes):  
[20, 25, 28, 30, 32, 35, 40, 45, 60]



## Step 2: Find the Median (Q2 → 50th Percentile)
- Median = middle value when data is sorted.  
- Here, 9 values → middle = 5th value.  
- Median = 32 minutes.  
Interpretation: Half of deliveries are faster than 32 minutes, half are slower.



## Step 3: Find Q1 (25th Percentile)
- Q1 = median of the **lower half** (below overall median).  
- Lower half = [20, 25, 28, 30]  
- Median of this half = average of 2nd and 3rd values = (25 + 28)/2 = 26.5 minutes.  
Interpretation: 25% of deliveries are faster than 26.5 minutes.



## Step 4: Find Q3 (75th Percentile)
- Q3 = median of the **upper half** (above overall median).  
- Upper half = [35, 40, 45, 60]  
- Median of this half = average of 2nd and 3rd values = (40 + 45)/2 = 42.5 minutes.  
Interpretation: 75% of deliveries are faster than 42.5 minutes.



## Step 5: Interquartile Range (IQR)
- IQR = Q3 − Q1 = 42.5 − 26.5 = 16 minutes.  
Interpretation: The middle 50% of deliveries vary within 16 minutes.



## Business Insight for Blinkit
- **Q1 (26.5 min)** → Fastest quarter of deliveries.  
- **Median (32 min)** → Typical delivery time.  
- **Q3 (42.5 min)** → Slowest quarter of deliveries.  
- **IQR (16 min)** → Spread of typical deliveries.  

Blinkit can use this to:  
- Promise “Most deliveries within 30–40 minutes.”  
- Identify outliers (very late deliveries beyond Q3).  
- Improve consistency by reducing variability in the upper quartile.
