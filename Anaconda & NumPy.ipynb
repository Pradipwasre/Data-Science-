## Installing Anaconda

Anaconda is a distribution of Python and R for scientific computing.  
It simplifies package management and deployment, especially for **Data Science** and **Machine Learning** projects.

## Why use Anaconda for Data Science?

- Comes with 1,500+ pre-installed packages (NumPy, Pandas, Matplotlib, Scikit-learn, etc.)
- Includes **Jupyter Notebook** for interactive coding
- Easy environment management
- Cross-platform (Windows, macOS, Linux)


## Downloading and Installing Anaconda

Steps:
1. Visit [https://www.anaconda.com/download](https://www.anaconda.com/download)

## Introduction to NumPy

NumPy = Numerical Python  
- Provides support for **multi-dimensional arrays**  
- Optimized for numerical computations  
- Essential for Data Science, ML, and Deep Learning


```python
pip install numpy
```

```python
import numpy as np
# Version
np.__version__
```

```python
# Creating Arrays
# 1D Array
arr1 = np.array([1, 2, 3, 4, 5])
```

```python
arr1
```

```python
# 2D Array
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
```

```python
arr2
```

```python
# 11. Array Attributes
print("Shape:", arr2.shape)
```

```python
# Data Type
arr2.dtype
```

```python
# Size
arr2.size
```

```python
# Dimensions:
arr2.ndim
```

# Type Conversion

```python

arr_float = arr1.astype(float)
print("Converted to float:", arr_float)
print("Data Type:", arr_float.dtype)
```

# 13. Simple Array Operations

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
```

```python
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

# 14. Accessing & Slicing Arrays

```python
print("First element:", arr1[0])
print("Slice [1:4]:", arr1[1:4])
```

```python
print("Row 1:", arr2[0])
print("Element [1,2]:", arr2[1,2])
```
